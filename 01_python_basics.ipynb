```python
name = 'data'
nums = [1, 2, 3, 4]
print(f'Hello {name}', sum(nums))
```

```python
def mean(xs):
    return sum(xs) / len(xs)

squares = [x**2 for x in nums]
print(mean(squares))
```

```python
person = {'name': 'Ada', 'age': 36}
for k, v in person.items():
    print(k, v)
```
